```python
import os
import sys
sys.path.append("../../")
from src.config_unitelma import CLASSIFIERS

from src.ml_pipeline import MLPipeline
from src.dl_pipeline import DLPipeline
import warnings
warnings.filterwarnings("ignore")

from dotenv import load_dotenv
load_dotenv()

TRAIN_PATH = os.path.join(os.getenv("UNITELMA_DATA"), "train_timeseries.csv")
TEST_PATH = os.path.join(os.getenv("UNITELMA_DATA"), "test_timeseries.csv")
OUTPUT_DIR = os.path.join(os.getenv("ROOT_DIR"), "notebooks_unitelma/runs/outputs")

LAGS_TO_TEST = [7, 14]
```

## Unitelma

```python
ml_processor = MLPipeline(
    train_path=TRAIN_PATH, 
    test_path=TEST_PATH, 
    output_path=OUTPUT_DIR, 
    use_macro=False
)

df_ml_results = ml_processor.run(
    lags=LAGS_TO_TEST,
    classifiers=CLASSIFIERS,
    apply_clipping=False,
    use_fs=False,
    time_aggregation='flatten'
)

display(df_ml_results)
```

```python
dl_processor = DLPipeline(
    train_path=TRAIN_PATH, 
    test_path=TEST_PATH, 
    output_path=OUTPUT_DIR, 
    use_macro=False
)

df_dl_results = dl_processor.run(
    lags=LAGS_TO_TEST,
    apply_clipping=False
)

display(df_dl_results)
```
